# Поиск фильма через OMDb API

Практическая работа 1 - репозитории и интеграции.

```python
import json
import os
import getpass
import requests
from dotenv import load_dotenv
```

## 1. Ввод названия фильма и API-ключа

```python
load_dotenv()
api_key = os.getenv("OMDB_API_KEY")

if not api_key:
    raise ValueError("Ключ OMDB_API_KEY не найден в .env")

title = input("Введите название фильма: ").strip()
```

## 2. Отправка GET-запроса

```python
response = requests.get(
    "https://www.omdbapi.com/",
    params={"t": title, "apikey": api_key},
    timeout=15
)
response.raise_for_status()
```

## 3. Получение и просмотр JSON

```python
json_data = response.json()
print(json.dumps(json_data, ensure_ascii=False, indent=2))
```

## 4. Вывод нужной информации

```python
if json_data.get("Response") == "False":
    print("Фильм не найден:", json_data.get("Error"))
else:
    movie_title = json_data.get("Title", "Нет данных")
    year = json_data.get("Year", "Нет данных")
    genre = json_data.get("Genre", "Нет данных")
    director = json_data.get("Director", "Нет данных")
    imdb_rating = json_data.get("imdbRating", "Нет данных")
    actors = json_data.get("Actors")

    print(f"Название: {movie_title}")
    print(f"Год: {year}")
    print(f"Жанр: {genre}")
    print(f"Режиссёр: {director}")
    print(f"Рейтинг IMDb: {imdb_rating}")
    print(f"Актёры: {actors}")
```
